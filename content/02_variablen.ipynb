## Operatoren

Zum Thema Operatoren könnte man ein eigenes Kapitel schreiben. Es gibt sehr viele Operatoren im Bezug auf Variablen und Wahrheitswerte, die einem das Leben erleichtern. Ich kann daher nur empfehlen auf den gängigen Webseiten zu Python zu forschen, was es da alles gibt.
Der gebräuchlichste Operator ist vermutlich das "+". Mit dem + können wir Zahlen zu einer Variable dazuaddieren oder wir können Texte erweitern. Beispiele:

```python
h = 8 + 12
i = "Das ist der erste Teil" + ", " + "das ist der zweite Teil"
print (h)
print(i)
```

Zum Rechnen können wir <b>"-", "*", "/"</b> verwenden. Zum Vergleichen benutzen wir <b>"=="</b>, da das einfache Gleichheitszeichen schon vergeben ist. Es gibt auch <b>"!="</b> (ungleich), <b>"<", "<=", ">", ">="</b>. Bei Vergleichen bekommen wir Wahrheitswerte (Boolean) als Ergebnis zurückgeliefert.

```python
print(30 == 25)
print(25 != 20)
```

Wir können manche Rechenoperation auch abkürzen, wenn wir z.B. nur den Wert einer Variablen erhöhen bzw. erniedrigen wollen:

```python
h += 5 # Variable h ist bereits weiter oben im Block definiert
print(h == 25)
h -= 10
print(h)
```

## Listen/Arrays

Listen sind eine Möglichkeit, um mehrere Werte oder Variablen zusammenhängend zu speichern. Sie funktionieren dabei wie Arrays. Die Besonderheit bei Python ist dabei, dass eine Liste unterschiedliche Datentypen aufnehmen kann. Es gibt viele Funktionen, die sich auf Listen ausführen lassen. Alle durchzugehen, würde zu weit führen, daher folgen ein paar Beispiele:

```python
autos = ["Mercedes", "Volvo", "BMW"]
anzahl_autos = len(autos) # gibt die Länge der Liste zurück
print(anzahl_autos)
autos.pop(0) # entfernt das erste Element (Index = 0) aus der Liste
autos.remove("Volvo") # entfernt das erste Element mit dem Wert "Volvo"
anzahl_autos = len(autos)
print(anzahl_autos)
autos.append("VW") # fügt ein neues Element ans Ende der Liste
autos.insert(1, "Ferrari") # fügt ein neues Element an eine bestimmte Stelle im Index
anzahl_autos = len(autos)
print(anzahl_autos)
autos[0] = "Porsche" # Verändere das Element an Stelle 0 (der Index eines Arrays beginnt immer bei 0)
print(autos)
```
