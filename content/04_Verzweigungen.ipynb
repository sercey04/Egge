## 📘 Abschnitt A – Einführung in Bedingungen

Eine Bedingung ist ein Ausdruck, der **True** oder **False** ergibt.

Beispiel:

```python
zahl = 5
print(zahl > 3)   # True
print(zahl == 10) # False
```

Eine einfache Verzweigung:

```python
if zahl > 3:
    print("Die Zahl ist größer als 3.")
else:
    print("Die Zahl ist nicht größer als 3.")
```

**Wichtig:**

- Einrückung ist Pflicht  
- Jede Verzweigung endet mit einem Doppelpunkt `:`
