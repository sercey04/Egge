# Schleifen (while-Schleife und for-Schleife)

Damit wir mehrfach auszuführenden Code nicht immer wieder neu aufschreiben müssen, können wir Anweisungen in Anweisungsblocks innerhalb sogenannter Schleifen zusammenfassen. Das hilft uns beispielsweise beim Zählen oder um für jedes Element einer Liste z.B. eine bestimmten Funktion auszuführen. 

## while-Schleife

while-Schleifen führen ihren Anweisungsblock solange aus bis die Eingangsbedingung nicht mehr erfüllt ist. Hierbei ist es wichtig, dass irgendwann die Eingangsbedingung <b>False</b> wird (oder ein break ausführen, dazu mehr später), weil wir sonst in einer Endlosschleife landen.
Hier ein einfaches Beispiel:

```python
i = 1
while i < 6:
    print(i)
    i += 1

print ("Die Schleife ist zuende")
```

Wenn du den Code ausgeführt hast, wirst du feststellen, dass er nach 5 Schluss gemacht hat. Das liegt daran, dass wir mit unserer Zählervariable <b>i</b> bei 1 gestartet sind. Dann geben wir in unserem Anweisungsblock <b>i</b> zunächst durch ein <b>print</b> in die Ausgabe aus. Letztlich erhöhen wir <b>i</b> bei jedem Schleifendurchlauf um 1. Sobald <b>i</b> den Wert 6 angenommen hat, ist die Bedingung der Schleife <b>False</b> und die Schleife ist dann beendet. Dann wird wieder der Code außerhalb der Schleife fortgesetzt.

### break
Wir haben auch die Möglichkeit einen Schleifendurchlauf durch das Keyword <b>break</b> zu beenden. Wir nehmen wieder das Beispiel von vorhin, geben jedoch mit Hilfe eines if-Statements ein break mit dazu:

```python
i = 1
while i < 6:
    print(i) 
    if ( i == 3 ): 
        break;
    i += 1

print ("Die Schleife ist zuende")
```

Wir sehen, jetzt wird unser Schleifendurchlauf bereits nach 3 beendet. Wir müssen uns sowohl bei der Schleifenbedingung als auch beim <b>break</b> darauf verlassen können, dass wir irgendwann die Schleife verlassen.

### continue
<b>continue</b> funktioniert ähnlich wie break, jedoch wird nicht die gesamte Schleife beendet, sondern nur der aktuelle Durchlauf:

```python
i = 0
while i < 6:
    i += 1
    if ( i == 3 or i == 6 ): 
        continue;
    print(i) 

print ("Die Schleife ist zuende")
```

Wir haben die Reihenfolge der Anweisungen im Beispiel ein wenig umgedreht. In der Ausgabe sehen wir, dass 3 und 6 nun fehlen, weil wir sie ausgeschlossen haben. Das <b>print(i)</b> wird nach dem <b>continue</b> für den Durchlauf nicht mehr ausgeführt.

## for-Schleife
Um Listen und andere Datentypen, welche Sammlungen von Werten sind (z.B. Tupel, Sets, Dictionaries, range, Strings), Wert für Wert abarbeiten zu können, können wir for-Schleifen benutzen. Dabei wird sequentiell, angefangen mit der ersten Stelle, jeder Wert in der Liste bzw. jeweiligen Sammlung betrachtet. Wir bleiben mal bei dem Beispiel einer Liste:

```python
fruechte = ["Apfel", "Banane", "Kirsche"]
for f in fruechte:
    print(f)
```

Wir können auch hier <b>break</b> und <b>continue</b> einsetzen:

```python
fruechte = ["Apfel", "Banane", "Kirsche"]
for f in fruechte:
    if (f == "Banane"):
        break;
    print(f)
```

```python
fruechte = ["Apfel", "Banane", "Kirsche"]
for f in fruechte:
    if (f == "Banane"):
        continue;
    print(f)
```
