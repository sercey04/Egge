# Projekte
Nachdem wir nun unsere Klausur geschrieben haben, haben wir noch ein paar Termine, um uns etwas freier und kreativer zu bewegen. Die Projekte sollen dazu da sein, dass jeder sich eigene Schwerpunkte setzen kann, an denen er arbeiten möchte. Idealerweise arbeitet ihr in 2-er Teams zusammen, um euch gegenseitig zu unterstützen. Sucht euch ein Thema aus den drei Vorschlägen heraus, was euch interessiert, und macht es zu eurem eigenen Projekt.

# Regeln
<ol>
<li><h3>Einsatz von KI:</h3><br>
Ihr dürft und solltet recherchieren, wie ihr bestimmte Probleme in Python und in der Entwicklungsumgebung (z.B. Jupyterlite Notebooks) umsetzen könnt. Wichtig dabei: Lasst nicht einfach die komplette Aufgabe von KI lösen, sondern arbeitet immer nur an einem bestimmten Problem, bei dem ihr gerade nicht weiterkommt. Ihr müsst außerdem die Lösung verstehen und in eigenen Worten erklären können!
</li>  
<li><h3>Entwicklungsumgebung:</h3><br>
Am sichersten ist, wenn ihr für euer Projekt hier mit den Jupyterlite Notebooks arbeitet. Über Imports können weitere Module mit erweiterten Funktionalitäten geladen werden (z.B. Grafiken, Visualisierung, Berechnungen). Ihr könnt auch z.B. mit WebTigerPython oder Trinket arbeiten, dann seid ihr aber auch verantwortlich dafür, dass eure Inhalte und Ergebnisse sicher gespeichert und gut zu präsentieren sind.
</li>  
<li><h3>Zwischenergebnisse:</h3><br>
Über die Pläne ladet ihr bitte eure Zwischenergebnisse hoch. Das dient einerseits der Ergebnissicherung, damit nichts verloren geht, und andererseits kann ich dann auch euren Fortschritt verfolgen.
</li>  
</ol> 

## 1. Idee: Casino
Simuliert Spiele aus dem Casino wie z.B. Roulette, Blackjack, Glücksrad, Slot-Machine, Poker etc. Das kann von einer einfachen Texteingabe bis hin zur grafischen Darstellung reichen. Erste Hürde ist es, die Regeln des Spiels in eine korrekte Programmlogik zu übersetzen.

Disclaimer: <b>Wir wetten nie um echtes Geld!</b>

Hier sind ein paar Vorschläge für Schwerpunkte und Erweiterungen:

<ol>
<li>Animationen für den Spielablauf</li> 
<li>Highscore-System (Bestenliste) über File oder Datenbank</li>    
<li>Statistiken und Wahrscheinlichkeits-Analyse z.B. mit dem math-Modul</li>  
<li>Fortgeschrittene Spielregeln wie "Double Down" und "Split" beim Blackjack</li>  
<li>Wett-Strategien (Logik für KI-Gegner schreiben, z.B. eine simple KI, die immer einen Teil ihres Geldes auf rot setzt)</li>  
<li>Login-System mit Kontostand-Speicherung</li>  
<li>Erweiterung in die Breite: Wahl des Spielers, welches Spiel er spielen möchte. Kontostand ggf. mitnehmen.</li>  
<li>Mehrspielermodus an einem Rechner</li>  
<li>+Deine eigenen Ideen!</li>  
</ol>

## 2. Idee: Passwort-Safe
Speichere Passwörter in einer Textdatei oder Datenbank. Setze dafür auch Verschlüsselungen wie die Caesar-Verschlüsselung ein.

Hier sind ein paar Vorschläge für Schwerpunkte und Erweiterungen:

<ol>
<li>Passwort-Stärke-Check (Validierung)</li>    
<li>Master-Passwort (Zugriffsschutz)</li>  
<li>Fortgeschrittene Verschlüsselung (Vigenère)</li>  
<li>Such- und Filterfunktion</li>  
<li>Button zum Kopieren eines ausgegebenen Passworts</li>
<li>Automatisches Ablaufdatum für Passwörter</li>  
<li>Export-Funktion (Backup) z.B. als CSV-Datei</li>  
<li>Grafische Oberfläche</li> 
<li>+Deine eigenen Ideen!</li>  
</ol>

## 3. Idee: Text- oder Grafikadventure
Denke dir eine Geschichte aus durch die du interaktiv führen möchtest. Du kannst beispielsweise mit Handlungssträngen arbeiten, die auseinander und ggf. wieder zusammenführen, oder du arbeitest mit dem Konzept der Räume. Vielleicht hast du auch selbst ein tolles Konzept!

Hier sind ein paar Vorschläge für Schwerpunkte und Erweiterungen:

<ol>
<li>Gegenstände/Inventar, die in bestimmten Situationen wichtig werden</li>
<li>Überlebens-System (z.B. HP, der Charakter kann durch Entscheidungen Lebenspunkte verlieren.)</li>    
<li>Rundenbasiertes Kampfsystem</li>  
<li>Zeitlimit oder Schrittzähler</li>  
<li>Rätsel</li>  
<li>Mini-Spiele</li>  
<li>Mehrere Enden (Multi-Ending)</li>  
<li>Speichersystem (Fortgeschritten)</li>
<li>+Deine eigenen Ideen!</li>  
</ol>
