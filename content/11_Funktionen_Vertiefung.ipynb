# Abschnitt A – Einführung: Funktionen in größeren Programmen

In *Notebook 10* haben Sie Funktionen definiert, Parameter übergeben und Rückgabewerte mit `return` genutzt.
Jetzt geht es darum, Funktionen so einzusetzen, dass Programme **übersichtlich** bleiben – auch wenn sie länger werden.

## 🔑 Neue Schwerpunkte

- **Mehrere Parameter**: Eine Funktion kann mehrere Werte verarbeiten.
- **Hilfsfunktionen**: Große Probleme werden in kleinere Funktionen zerlegt.
- **Funktionen + Daten**: Funktionen werten Listen aus oder erzeugen neue Listen.

## ▶️ Beispiel: Mehrere Parameter + Rückgabe

```python
def rechteck_flaeche(breite, hoehe):
    return breite * hoehe

f = rechteck_flaeche(3, 5)
print(f)  # 15
```

Merke: Eine Funktion sollte eine **klare Aufgabe** haben. Das Hauptprogramm ruft die Funktionen auf und kombiniert die Ergebnisse.

# 🧾 Zusammenfassung dieses Notebooks

## 🎯 Wichtige Inhalte
- Funktionen können mehrere Parameter verarbeiten.
- Schleifen lassen sich sinnvoll in Funktionen kapseln.
- Listen lassen sich mit Funktionen auswerten (Summe, Durchschnitt, Minimum/Maximum).
- Typische Fehler: fehlendes `return`, falsche Parameter, Funktion nicht aufgerufen.
- Größere Programme werden übersichtlich, wenn sie aus mehreren Funktionen bestehen.

📚 Glossar
- **Hilfsfunktion**: kleine Funktion, die einen Teil eines Problems löst.
- **Parameter**: Platzhalter in der Funktionsdefinition.
- **Argument**: konkreter Wert beim Funktionsaufruf.
- **Rückgabewert**: Wert, den eine Funktion mit `return` liefert.
- **None**: Python-Wert, der oft erscheint, wenn `return` fehlt.
