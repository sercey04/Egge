# Funktionen (Functions)

Funktionen, oder in manchen Programmiersprachen auch Prozeduren genannt, helfen uns Befehle oder Berechnungen sinnvoll zu strukturieren. Wir können die Funktionen dann an verschiedenen Stellen im Code ausführen, ohne den Code jedes Mal neu schreiben zu müssen. Wenn wir an die letzte Übung zum Thema Variablen denken, dann wollten wir an einer Stelle die Umrechnung von Celsius zu Fahrenheit vornehmen. Anstatt jedes Mal die Formel neu aufzuschreiben, wenn wir eine Umrechnung machen, kann uns eine einzige Funktion aus dem Celsius-Wert als Ergebnis den Fahrenheit-Wert liefern. Dazu später mehr. Das einfachste Beispiel für eine Funktion ist die Ausgabe von "Hello World".

```python
def hello_world():
  print("Hello world")

hello_world()
```

Wir definieren eine Funktion mit dem <b>def</b>-Keyword. Danach folgt eine Bezeichnung, die wir nach den selben Regeln wie bei den Variablen wählen können (siehe den gelben Kasten in 02_Variablen). Dann folgen runde Klammern, in die wir auch noch Parameter hineinschreiben können, das werden wir später sehen. Zu guter letzt folgt noch ein Doppelpunkt und dann folgen die Anweisungen ab der nächsten Zeile eingerückt.
Schließlich können wir die Funktion mit unserem gewählten Namen und wieder den runden Klammern aufrufen. Das können wir auch mehrmals hintereinander machen.

```python
hello_world()
hello_world()
hello_world()
```

## Rückgabewerte (Return value)

Funktionen können nicht nur Befehle ausführen, sondern sie können auch einen Wert zurückgeben. Der Datentyp passt sich, wenn nicht explizit definiert, wie bei den Variablen den mitgelieferten Werten an. Wir können die hello_world Funktion so umschreiben, dass sie nicht direkt das print ausführt, sondern stattdessen geben wir den String als Rückgabewert an. Wichtig dabei ist das <b>return</b>-Keyword


```python
def hello_world2():
    return "Hello world 2"
# jetzt schreiben wir den Funktions-Aufruf in das print
print(hello_world2())
```

Damit können wir z.B. eine Funktion schreiben, welche uns den aktuellen Tag als Rückgabewert zurückgibt.

```python
from datetime import date # damit wir date nutzen können, müssen wir es zuerst in unseren Code importieren

def aktueller_tag():
    return date.today() # date.today() ist auch eine Python-Funktion, welche schon in der datetime-Bibliothek von Python vordefiniert ist
# Und wieder geben wir die Funktion zum Überpüfen aus
print(aktueller_tag())    
```

## Parameter (Arguments)

Nun haben wir bisher immer nur feste Werte zurückgeben können, da wir der Funktion noch nichts mitgeliefert haben, was sie verarbeiten könnte. Dafür können wir Parameter einer Funktion definieren. Diese werden in die runden Klammern geschrieben und zeigen an, dass die Funktion einen oder mehrere Parameter erwartet. Nehmen wir wieder unser Beispiel für die Umrechnung der Temperatur von Celsius zu Fahrenheit:

```python
def celsius_zu_fahrenheit(celsius):
    return (celsius * 9/5) + 32

def fahrenheit_zu_celsius(fahrenheit):
    return (fahrenheit - 32) * 5 / 9  

fahrenheit = celsius_zu_fahrenheit(25); # wir rufen die Funktion mit dem Wert 25 auf, wir speichern unser Ergebnis zwischen, damit wir es später erneut nutzen können
print(fahrenheit) # jetzt geben wir den Fahrenheit-Wert aus
print(fahrenheit_zu_celsius(fahrenheit)) # wir machen noch die Gegenprobe, ob wir wieder den korrekten Celsius-Wert zurückbekommen, den wir eingegeben haben
```

Wir können das gleiche Prinzip nutzen, um z.B. Textbausteine hinzuzufügen. Wir geben dieses Mal 3 Parameter vor, für einen setzen wir jedoch einen default, einen Standardwert. Wir nutzen wieder die f-Schreibweise, um die Parameter in unseren String einfügen zu können.

```python
def email_vorlage(adressat, betreff, gruss = "Mit freundlichen Grüßen"):
    return f"Hallo {adressat}. Es geht um {betreff}. Ich bitte um Antwort. {gruss}"

print(email_vorlage(gruss = "Viele Grüße", adressat = "Herr Müller", betreff = "die Uhrzeit für das heutige Meeting")) # hier haben wir explizit die Namen der Parameter angegeben, so weiß die Funktion, wie sie welchen Wert einsetzen muss
print(email_vorlage("Herr Müller", "die Uhrzeit für das heutige Meeting", "Viele Grüße")) # wir brauchen die Parameter nicht benennen, wenn wir die richtige Reihenfolge einhalten
print(email_vorlage("Herr Müller", "die Uhrzeit für das heutige Meeting")) # falls wir für den letzten Wert nichts angeben, wird der Standard-Wert verwendet
```
