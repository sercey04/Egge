## <b>Grafik-Adventure mit Textabfrage</b>

```python
from IPython.display import Image, display, clear_output

# Hilfsfunktion zum Anzeigen von Bildern
def zeige_szene(bild_url, text):
    clear_output(wait=True) # Löscht die alte Szene für einen sauberen Look
    display(Image(filename=bild_url, width=600))
    print(text)

def start_spiel():
    burg_url = "assets/burg.png" # Greift auf das Image im lokalen Pfad zu
    
    zeige_szene(burg_url, "--- DAS GEHEIMNIS DER ALTEN BURG ---\nDu stehst vor einem schweren Eichentor.")
    wahl = input("Möchtest du das Tor 'oeffnen' oder 'umkehren'? > ").lower()

    if wahl == "oeffnen":
        halle()
    else:
        print("\nDas Abenteuer endet hier.")

def halle():
    halle_url = "assets/halle.png" # Greift auf das Image im lokalen Pfad zu
    zeige_szene(halle_url, "\nDu betrittst eine dunkle Halle mit zwei Gängen.")
    wahl = input("Gehst du nach 'links' oder 'rechts'? > ").lower()
    
    if wahl == "links":
        print("\nDu hast den Schatz gefunden! SIEG!")
    else:
        print("\nEine Sackgasse...")

# Spiel starten
start_spiel()
```

## <b>Grafik mit Button</b>

```python
import piplite
await piplite.install('ipycanvas')

import ipywidgets as widgets
from IPython.display import display

# 1. Bild-Widget erstellen (lokale Datei oder URL)
file = open("assets/burg.png", "rb")
image_data = file.read()
image_widget = widgets.Image(
    value=image_data,
    format='png',
    width=600
)

# 2. Button erstellen
button = widgets.Button(description="Bild-Aktion auslösen")
output = widgets.Output()

# 3. Klick-Funktion
def on_click(b):
    with output:
        output.clear_output() # Optionale Reinigung des Bereichs
        print("Text unter dem Bild!")

button.on_click(on_click)

# 4. Zusammen in einer VBox (vertikal) anzeigen
layout = widgets.VBox([image_widget, button, output])
display(layout)
```

## <b>Grafik mit Click-Event</b>

```python
import piplite
await piplite.install('ipycanvas')
await piplite.install('ipyevents')

import ipywidgets as widgets
from ipyevents import Event
from IPython.display import display

# 1. Bild laden
image = widgets.Image(value=open("assets/burg.png", "rb").read(), format='png', width=600)

# 2. Event-Watcher für das Bild erstellen
# 'click' reagiert auf Mausklicks
img_events = Event(source=image, watched_events=['click'])
output = widgets.Output()

# 3. Callback-Funktion (enthält Koordinaten und Event-Details)
def handle_click(event):
    with output:
        output.clear_output()
        # event['relativeX'] und event['relativeY'] geben die Klick-Position an
        print(f"Bild angeklickt bei: X={event['relativeX']}, Y={event['relativeY']}")

img_events.on_dom_event(handle_click)

display(image, output)
```
