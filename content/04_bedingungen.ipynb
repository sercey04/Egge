# Bedingungen (if-else und match)

In unseren Funktionen und unserer Programmlogik wollen wir die Möglichkeit haben, unser Vorgehen an z.B. bestimmte Eingabewerte einer Variablen anzupassen und die jeweils passenden Zeilen Code auszuführen.

So kann es z.B. einen Unterschied machen, was für eine Ausgabe eine Nutzer:in angezeigt bekommt, je nachdem, ob sie volljährig oder nicht ist.

Dafür brauchen wir eine Möglichkeit, Werte überprüfen oder vergleichen zu können.Vergleichsoperatoren helfen uns, den Wahrheitgehalt einer Aussage auszuwerten. Folgende Vergleichsoperatoren sind gängig:

<ul>
    <li><b>==</b>: Überprüft auf Gleichheit. True bei Gleichheit, sonst False. 
    <li><b>!=</b>: Überprüft auf Ungleichheit. False bei Gleichheit, sonst True.
    <li><b>>, &lt </b>: Überprüft auf echt größer/kleiner. True bei echt größer/kleiner, sonst False.   
    <li><b>>=, &lt=</b>: Überprüft auf größer/kleiner oder Gleichheit. True bei größer/kleiner und Gleichheit, sonst False.</li>
</ul>

Wir bekommen also einen bool'schen Wert als Ergebnis zurück (vgl. Datentyp Boolean bei Variablen). Diesen können wir uns auch in einer Ausgabe wiedergeben lassen.

```python
print(23 > 42)
```

## if 
Um daraus jetzt eine Bedingung machen zu können, benötigen wir noch ein keyword. Die häufigste Variante ist dabei <b>if</b>. Wir schreiben <i>if Wert1 Vergleichsoperator Wert2:<i/> Danach werden die Anweisungen, ähnlich wie bei den Funktionen, eingerückt aufgeschrieben.

```python
alter = 18
if alter >= 18:
    print("Du bist volljährig.")
```

## else
Mit dem keyword <b>else</b> in Verbindung mit dem if, können wir Anweisungen geben, die ausgeführt werden sollen, wenn die Bedingung nicht erfüllt ist.

```python
alter = 15
if alter >= 18:
    print("Du bist volljährig.")
else:
    print("Du bist noch minderjährig")
```

```python
fuehrerschein = True
if fuehrerschein == True:
print("Du darfst (ggf. in Begleitung) Auto fahren.")
else:
print("Du darst nicht selber Auto fahren.")
```

## elif
Neben else können wir auch noch <b>elif</b> (kurz für else if) definieren. Damit können wir eine weitere Bedingung formulieren, falls wir mehr als eine Bedingung überprüfen wollen. Wir können dies auch wieder mit else kombinieren. Dabei ist zu beachten: Sobald die erste Bedingung erfüllt wird, werden die nachfolgenden Bedingungen nicht mehr überprüft!

```python
punktzahl = 99
if punktzahl >= 85:
    print("Du hast eine sehr gute Leistung vollbracht.")
elif punktzahl >= 70:
    print("Du hast eine gute Leistung vollbracht.")
elif punktzahl >= 50:
    print("Du hast eine ordentliche Leistung vollbracht.")
else:
    print("Du warst auch dabei.")
```

## Logische Operatoren

Neben unseren Vergleichsoperatoren haben wir noch logische Operatoren, die eine Verkettung von Bedingungen ermöglichen.

Diese wichtigsten lauten: 
<ul><li><b>and</b>: Alle mit and verketteten Bedingungen müssen erfüllt sein.</li> <li><b>or</b>: Eine der mit or verketteten Bedingungen muss nur erfüllt sein.</li> <li><b>not</b>: Die Bedingung wird negiert. D.h. das Gegenteil wird überprüft.</li>.</ul>

Wichtig hierbei ist zu beachten, dass bei Verkettung von verschiedenen Operatoren ggf. Klammern gesetzt werden müssen, um das gewünschte Ergebnis zu erhalten.

Beispiel von einer einfachen Verkettung:


```python
punktzahl = 85
if punktzahl <= 100 and punktzahl >= 0:
    print("Das ist eine gültige Punktzahl.")
else:
    print("Das ist keine gültige Punktzahl.")
```

## Match

Für Fälle in denen wir lange Aneinanderreihungen von elif Anweisungen schreiben müssten, gibt es eine weitere Variante, Code nach bestimmten Bedingungen ausführen zu lassen. Diese wird mit dem keyword <b>match</b> definiert. Dann wird wieder eingerückt und Fall wird mit dem keyword <b>case</b> gefolgt von dem zu überprüfenden Fall eingeleitet. Dann wird erneut eingerückt und Anweisungen geschrieben. Es kann manchmal sinnvoll sein einen Standardfall mit <b>case _:</b> zu deklarieren, falls etwas ausgeführt werden sollte, wenn der zu überprüfende Wert in keiner der Fälle vorkommt.

Beispiel:

```python
befehl = "start"
match befehl:
    case "start":
        print("System wird hochgefahren...")
    case "stop":
        print("System wird abgeschaltet.")
    case _: # Wildcard/Default-Fall
        print("Unbekannter Befehl.")
```

Das ist besonders interessant für Unterteilungen in z.B. Wochentage, Monate oder vordefinierte Wertebereiche (z.B. 0-50, 51-100, 100-150 ...), also wenn wir bereits mehrere zu erwartenden Werte kennen. Wir können auch Fälle zusammenfassen, indem wir das Zeichen <b>|</b> zwischen zwei Werte setzen:

```python
wochentag = "donnerstag"
match wochentag:
    case "montag":
        print("Heute wird hart")
    case "dienstag"|"donnerstag": # hier haben wir zwei Fälle zusammengefasst
        print("Heute ist ganz ok.")
    case "mittwoch":
        print("Heute bin ich motiviert, Hälfte der Woche rum")
    case "freitag":
        print("Bald Wochenende. :)")
    case "samstag"|"sonntag": # hier auch
        print("Yay, Wochenende!")
    case _: # Wildcard/Default-Fall
        print("Wochentag nicht gefunden, evtl. Schreibweise prüfen")
```
