## 📘 Abschnitt A – Zellen & Reihenfolge

In Jupyter werden Zellen **nacheinander** ausgeführt.  
Variablen existieren erst, nachdem der Code in einer Zelle ausgeführt wurde.

Führen Sie die Zellen am besten der **Reihenfolge nach** aus (von oben nach unten).


````python
# Beispiel: zuerst diese Zelle ausführen
x = 10
print("x wurde gesetzt.")
````

````python
# Danach diese Zelle ausführen
print("Der Wert von x ist:", x)
````

Wenn Sie eine Zelle ausführen, in der eine Variable verwendet wird,  
bevor sie gesetzt wurde, entsteht ein `NameError`.

## 📚 Glossar

- **NameError** – eine Variable oder ein Name ist unbekannt (nicht definiert oder falsch geschrieben).  
- **SyntaxError** – der Code ist formal falsch aufgebaut (z. B. fehlende Klammer, Anführungszeichen).  
- **TypeError** – unpassende Datentypen werden kombiniert (z. B. Zahl + Text ohne Umwandlung).  
- **Debugging** – systematisches Suchen und Beheben von Fehlern im Programm.  
- **Testzelle** – eine separate Zelle, in der ein Teilprogramm gezielt ausprobiert wird.  
- **Ausführungsreihenfolge** – die Reihenfolge, in der Zellen ausgeführt werden; sie beeinflusst, welche Variablen wann verfügbar sind.
