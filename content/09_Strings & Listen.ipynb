📚 Glossar
- **Token**: Elementare Einheit eines Texts (z. B. Wort).
- **Trennzeichen**: Zeichen, die Einheiten voneinander trennen (z. B. `,` oder Leerzeichen).
- **Normierung**: Vereinheitlichung (z. B. Kleinschreibung).
- **CSV**: Comma Separated Values – einfache Textdatenstruktur.
- **Frequenz**: Anzahl der Vorkommen.
