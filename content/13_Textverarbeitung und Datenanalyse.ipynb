# Abschnitt A – Einführung: Von Text zu Daten

Texte enthalten viele Informationen. Damit ein Computer sie auswerten kann, müssen wir sie schrittweise
so verändern, dass sie für den Rechner gut handhabbar werden.

Typische Schritte dabei sind:
- Groß- und Kleinschreibung vereinheitlichen
- störende Zeichen wie Satzzeichen entfernen
- Texte in einzelne Wörter zerlegen

Diese Schritte sind keine formalen Fachbegriffe, sondern beschreiben einfach, **was wir mit dem Text tun**, 
bevor wir ihn auswerten.

## ▶️ Beispiel: Wörter zählen

```python
text = "Hallo Welt! Hallo Informatik."
t = text.lower()
t = t.replace("!", "").replace(".", "")
woerter = t.split()

haeufig = {}
for w in woerter:
    if w in haeufig:
        haeufig[w] = haeufig[w] + 1
    else:
        haeufig[w] = 1

print(haeufig)  # {'hallo': 2, 'welt': 1, 'informatik': 1}
```

## 🎯 Wichtige Inhalte

- `split()` zerlegt Texte in Wörter (Tokens).

- Textanalyse ist ein Zusammenspiel aus Strings, Listen und Dictionaries.

📚 Glossar
- **Normierung**: Vereinheitlichung (z. B. Kleinschreibung).
- **Bereinigung**: Entfernen/Ersetzen störender Zeichen.
- **Token**: Einheit eines Textes (z. B. Wort).
- **Frequenz**: Anzahl der Vorkommen.
- **Stopwort**: häufiges Wort, das oft ignoriert wird (z. B. „und“).
