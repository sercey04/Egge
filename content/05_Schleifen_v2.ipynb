# Abschnitt A – Einführung: Schleifen

In vielen Programmen sollen Anweisungen **mehrfach wiederholt** werden.
Dafür stellt Python sogenannte **Schleifen** zur Verfügung.

Eine Schleife führt einen Codeblock **so lange oder so oft aus**, wie eine bestimmte Bedingung erfüllt ist
oder eine vorgegebene Anzahl erreicht wurde.

## 🔑 Neue Begriffe

- **Schleife**: Programmstruktur zur Wiederholung von Anweisungen.
- **Iteration**: ein einzelner Durchlauf einer Schleife.
- **Schleifenrumpf**: der eingerückte Codeblock, der wiederholt ausgeführt wird.

In Python unterscheiden wir zwei wichtige Schleifenarten:
- die **for-Schleife** (feste Anzahl von Wiederholungen)
- die **while-Schleife** (Wiederholung solange eine Bedingung gilt)

## 🧱 Aufbau einer for-Schleife

```python
for i in range(5):
    print(i)
```

Erläuterung:
- `range(5)` erzeugt die Zahlen 0 bis 4.
- `i` nimmt diese Werte nacheinander an.
- Der eingerückte Block wird für jeden Wert einmal ausgeführt.

## 🧱 Aufbau einer while-Schleife

```python
zaehler = 0
while zaehler < 5:
    print(zaehler)
    zaehler = zaehler + 1
```

Erläuterung:
- Die Schleife läuft, **solange die Bedingung wahr ist**.
- Achtung: Die Bedingung muss sich ändern, sonst entsteht eine Endlosschleife.

📚 Glossar
- **Schleife**: Wiederholt Anweisungen mehrfach.
- **Zählschleife**: Schleife mit fester Wiederholungszahl (meist `for` + `range`).
- **Bedingte Schleife**: Wiederholt solange eine Bedingung erfüllt ist (`while`).
- **Akkumulator**: Variable, die Werte über Wiederholungen sammelt (z. B. Summe).
- **Abbruchbedingung**: Kriterium, das die Wiederholung beendet.
