📚 Glossar
- **Index**: Position eines Zeichens im String.
- **Slice**: Teilstück eines Strings (`text[a:b]`).
- **Methode**: Funktion, die an einem Objekt aufgerufen wird.
- **Palindrom**: Wort/Satz, der vorwärts wie rückwärts gleich ist.
- **Tokenisierung**: Zerlegung eines Texts in Bestandteile (z. B. Wörter).
