## 📚 Glossar

- **Variable** – Name für einen gespeicherten Wert.  
- **Datentyp** – beschreibt die Art eines Wertes (z. B. Zahl, Text).  
- **`int`** – Ganzzahl (z. B. 3, 42).  
- **`float`** – Kommazahl (z. B. 3.14).  
- **`str`** – Text (String), steht in Anführungszeichen.  
- **`type()`** – Funktion, die den Datentyp eines Wertes anzeigt.  
- **`TypeError`** – Fehler bei unpassenden Datentyp-Kombinationen.  
- **`NameError`** – Fehler, wenn eine Variable noch nicht existiert oder falsch geschrieben ist.
