# Abschnitt A – Einführung: Was ist ein Dictionary?

Bisher haben Sie vor allem mit **Listen** gearbeitet (vgl. *Notebook 08 – Listen*). Listen eignen sich gut, wenn Werte eine feste Reihenfolge haben und man sie per Index (`0, 1, 2, …`) anspricht.

Ein **Dictionary** (auch „Wörterbuch“) speichert Daten anders:
- statt Indexen gibt es **Schlüssel** (keys)
- zu jedem Schlüssel gehört ein **Wert** (value)

Damit können Sie Werte **gezielt** über einen passenden Namen abrufen, z. B. eine Note zu einem Namen oder eine Häufigkeit zu einem Wort.

## 🔑 Neue Begriffe

- **Dictionary (dict)**: Datenstruktur aus **Schlüssel–Wert-Paaren**.
- **Schlüssel (key)**: eindeutige „Adresse“ im Dictionary.
- **Wert (value)**: der zugehörige Inhalt.
- **Schlüssel–Wert-Paar**: Kombination aus key und value, z. B. `"Anna": 2`.

## 🧱 Aufbau der Syntax

Ein Dictionary wird mit `{}` erstellt:

```python
noten = {"Anna": 2, "Ben": 3}
```

Zugriff erfolgt über den Schlüssel:

```python
print(noten["Anna"])  # 2
```

## ▶️ Beispiel: Häufigkeiten zählen (klassischer Einsatz)

```python
woerter = ["apfel", "birne", "apfel"]
haeufig = {}

for w in woerter:
    if w in haeufig:
        haeufig[w] = haeufig[w] + 1
    else:
        haeufig[w] = 1

print(haeufig)  # {'apfel': 2, 'birne': 1}
```

Merke: Dictionaries sind besonders nützlich, wenn Sie **zuordnen** oder **zählen** möchten.

📚 Glossar
- **Dictionary**: Datenstruktur aus Schlüssel–Wert-Paaren.
- **Schlüssel (key)**: eindeutige Adresse im Dictionary.
- **Wert (value)**: zugehöriger Inhalt.
- **Schlüssel–Wert-Paar**: Kombination aus key und value.
- **Häufigkeit**: Anzahl der Vorkommen eines Elements.
