## 📚 Glossar

- **Ausgabe** – Text oder Wert, der unter einer Code-Zelle erscheint.  
- **Kommentar** – beginnt mit `#` und wird von Python nicht ausgeführt; dient zur Erklärung und Strukturierung.  
- **Fehler / Exception** – Meldung von Python, dass der Code nicht ausgeführt werden kann.  
- **NameError** – ein Name ist unbekannt, häufig durch fehlende Anführungszeichen bei Text.  
- **SyntaxError** – der Code ist syntaktisch ungültig (z. B. fehlende Klammer oder Anführungszeichen).  
- **ASCII-Art** – einfache „Grafik“ aus Zeichen (z. B. `*`, `#`, `-`).  
- **Notebook-Zelle** – ein Block im Notebook, der entweder Code oder Text enthält.
