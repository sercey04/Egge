# Abschnitt A – Einführung: Fehler abfangen mit `try` / `except`

Sie kennen bereits viele Fehlersituationen aus *Notebook 03 – Zellen, Variablen und Fehlersuche*.
Neu ist jetzt ein Werkzeug, mit dem Programme **nicht sofort abbrechen**, wenn etwas schiefgeht.

## ▶️ Beispiel: Programm bricht nicht mehr ab

```python
try:
    zahl = int(input("Zahl: "))
    print("Ergebnis:", 10 / zahl)
except:
    print("Eingabe ungültig oder Division durch 0.")
```

**Beobachtung:**
- Bei korrekter Eingabe läuft das Programm normal.
- Bei falscher Eingabe erscheint eine Meldung, das Programm stürzt aber **nicht** ab.

## 🔑 Wichtige Begriffe (am Beispiel erklärt)

- **Exception**: Fehler, der während der Programmausführung auftritt  
  (z. B. falsche Eingabe, Division durch 0).
- **try-Block**: Code, der *ausprobiert* wird und fehlschlagen kann.
- **except-Block**: Code, der ausgeführt wird, **wenn ein Fehler auftritt**.

Merke: `try/except` ersetzt keine Fehlersuche, sondern hilft dabei,
Programme **robuster** zu machen.

📚 Glossar

- **try/except**: Konstruktion zum Abfangen von Exceptions.
- **Logikfehler**: Programm läuft, Ergebnis ist aber falsch.
- **Debugging**: systematisches Suchen und Beheben von Fehlern.
