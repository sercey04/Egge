📚 Glossar
- **Operator**: Symbol/Schlüsselwort für eine Operation.
- **Verzweigung**: Auswahl von Fällen (`if`/`elif`/`else`).
- **Iteration**: Wiederholung von Anweisungen.
- **Akkumulator**: Sammelvariable (z. B. Summe).
- **Menü**: wiederkehrende Benutzerinteraktion mit Auswahl.
