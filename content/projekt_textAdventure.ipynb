## <b>Text-Adventure mit Räumen und Piratensetting</b>

```python
import piplite
await piplite.install('ipycanvas')

import ipywidgets as widgets
from IPython.display import display, clear_output

# --- Spielzustand (State) ---
state = {
    "raum": "hafen",
    "inventar": [],
    "rum_getrunken": False
}

# --- UI Elemente ---
text_ausgabe = widgets.Output(layout={'border': '1px solid black', 'height': '200px', 'overflow_y': 'scroll'})
button_container = widgets.HBox()

def schreibe(text):
    with text_ausgabe:
        print(text)

def update_buttons(optionen):
    """Erzeugt Knöpfe basierend auf den Möglichkeiten im Raum."""
    buttons = []
    for label, ziel in optionen.items():
        btn = widgets.Button(description=label)
        btn.on_click(lambda b, z=ziel: gehe_zu_raum(z))
        buttons.append(btn)
    button_container.children = buttons

# --- Raum-Funktionen ---
def gehe_zu_raum(raum_name):
    state["raum"] = raum_name
    with text_ausgabe:
        clear_output()
    
    if raum_name == "hafen":
        schreibe("Jack Sparrow steht am Hafen von Tortuga. Die 'Black Pearl' ist weg!")
        schreibe("Ein betrunkener Pirat starrt dich an.")
        update_buttons({
            "Pirat ansprechen": "pirat",
            "In die Taverne gehen": "taverne"
        })

    elif raum_name == "pirat":
        schreibe("'Aye Jack! Suchst du dein Schiff? Der Hafenmeister hat den Schlüssel.'")
        if "Kompass" not in state["inventar"]:
            schreibe("Er gibt dir deinen Kompass zurück.")
            state["inventar"].append("Kompass")
        update_buttons({"Zurück zum Hafen": "hafen"})

    elif raum_name == "taverne":
        schreibe("Es riecht nach altem Fisch und Freiheit. Der Wirt putzt ein Glas.")
        optionen = {"Hafenmeister suchen": "hafenmeister", "Zurück zum Hafen": "hafen"}
        if not state["rum_getrunken"]:
            optionen["Ein Glas Rum trinken"] = "trinken"
        update_buttons(optionen)

    elif raum_name == "trinken":
        state["rum_getrunken"] = True
        schreibe("Jack: 'Klar soweit!' Du fühlst dich mutiger, aber die Sicht wird schlechter.")
        update_buttons({"Weiter geht's": "taverne"})

    elif raum_name == "hafenmeister":
        if "Kompass" in state["inventar"]:
            schreibe("Mit dem Kompass hast du den Meister im Hinterzimmer gefunden!")
            schreibe("Er zittert: 'Hier Jack, nimm die Schlüssel für das Ruderboot!'")
            schreibe("--- DU HAST GEWONNEN ---")
            update_buttons({"Neustart": "hafen"})
        else:
            schreibe("Du findest den Hafenmeister nicht. Vielleicht kann dir jemand am Hafen helfen?")
            update_buttons({"Zurück zur Taverne": "taverne"})

# --- Spielstart ---
display(text_ausgabe, button_container)
gehe_zu_raum("hafen")
```
