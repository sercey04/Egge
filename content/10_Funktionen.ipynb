# Abschnitt A – Einführung: Was sind Funktionen?

Bisher bestanden Ihre Programme aus einer fortlaufenden Abfolge von Anweisungen. Mit **Funktionen** können Sie zusammengehörige Anweisungen **bündeln**, **benennen** und **mehrfach verwenden** (vgl. *Notebook 05 – Schleifen*: Wiederholung von Abläufen).

Eine Funktion ist ein **eigener Programmbaustein** mit einer klaren Aufgabe.

## 🔑 Neue Begriffe

- **Funktion**: benannter Programmteil mit einer Aufgabe.  
- **Funktionsdefinition**: die Stelle, an der festgelegt wird, *was* die Funktion tut.  
- **Funktionsaufruf**: die Nutzung der Funktion an einer anderen Stelle im Programm.  
- **Parameter**: Platzhalter für Werte, die beim Aufruf übergeben werden.

## 🧱 Aufbau der Syntax

Eine Funktion wird in Python mit `def` definiert:

```python
def funktionsname():
    # Anweisungen der Funktion
```

Wichtig:
- Nach dem Funktionsnamen stehen immer **Klammern** `()`.
- Danach kommt ein **Doppelpunkt** `:`.
- Der Funktionskörper ist **eingerückt** (wie bei Verzweigungen/Schleifen).

## ▶️ Beispiel

```python
def begruesse():
    print("Hallo!")

begruesse()  # Funktionsaufruf
```

Merke: Eine Funktion wird **erst ausgeführt**, wenn sie **aufgerufen** wird.

# Abschnitt C – Rückgabewerte mit `return`

Bisher haben Ihre Funktionen Ergebnisse **nur ausgegeben** (`print`). Häufig ist es aber sinnvoller, ein Ergebnis **zurückzugeben**, damit es im Programm weiterverarbeitet werden kann.

- `print(...)` zeigt etwas auf dem Bildschirm.  
- `return ...` gibt einen Wert an die Stelle zurück, an der die Funktion aufgerufen wurde.

## ▶️ Beispiel

```python
def quadratzahl(n):
    return n * n

q = quadratzahl(5)
print(q)  # 25
```

# 🧾 Zusammenfassung dieses Notebooks

## 🎯 Wichtige Inhalte
- Funktionen strukturieren Programme und machen Code wiederverwendbar.
- Eine Funktionsdefinition beginnt mit `def`, enthält `()` und endet mit `:`.
- Parameter machen Funktionen flexibel.
- `return` liefert Werte zurück, `print` dient der Ausgabe.
- Rückgabewerte lassen sich speichern und weiterverarbeiten.

📚 Glossar
- **Funktion**: benannter Programmbaustein mit eigener Aufgabe.
- **Funktionsdefinition**: Beschreibung, *was* die Funktion tut (`def ...`).
- **Funktionsaufruf**: Nutzung einer Funktion im Programm (z. B. `f(3)`).
- **Parameter**: Platzhalter in der Funktion, der beim Aufruf ersetzt wird.
- **Argument**: konkreter Wert beim Aufruf einer Funktion.
- **Rückgabewert**: Ergebnis, das eine Funktion mit `return` zurückliefert.
