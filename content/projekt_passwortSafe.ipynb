## <b>Passwort-Safe mit Textdatei</b>

```python
import piplite
await piplite.install('ipycanvas')

import ipywidgets as widgets
from IPython.display import display
import random
import string

# Widgets erstellen
service_input = widgets.Text(description="Dienst:", placeholder="z.B. Instagram")
length_slider = widgets.IntSlider(value=12, min=8, max=32, description="Länge:")
generate_btn = widgets.Button(description="Generieren & Speichern", button_style='success')
output = widgets.Output()

display(service_input, length_slider, generate_btn, output)

def generate_password(length):
    chars = string.ascii_letters + string.digits + string.punctuation
    return ''.join(random.choice(chars) for _ in range(length))

def save_to_file(service, password):
    # Einfache Cäsar-Verschlüsselung (Erweiterung)
    encrypted = "".join(chr(ord(c) + 3) for c in password)
    
    with open("passwords.txt", "a") as f:
        f.write(f"{service}: {encrypted}\n")

def on_button_clicked(b):
    with output:
        output.clear_output()
        pw = generate_password(length_slider.value)
        save_to_file(service_input.value, pw)
        print(f"✅ Passwort für {service_input.value} gespeichert!")
        print(f"Original (nur zur Info): {pw}")

generate_btn.on_click(on_button_clicked)
```

## <b>Passwort-Safe mit Datenbank</b>

```python
import piplite
await piplite.install('tinydb')

from tinydb import TinyDB, Query
from tinydb.storages import MemoryStorage

# 1. Datenbank im Arbeitsspeicher initialisieren (KEINE Datei-Fehler möglich)
db = TinyDB(storage=MemoryStorage)

# 2. Daten einfügen
db.insert({'dienst': 'Netflix', 'passwort': 'geheim_verschluesselt'})
db.insert({'dienst': 'Instagram', 'passwort': 'mein_sicheres_pw'})

# 3. Test-Abfrage
User = Query()
ergebnis = db.search(User.dienst == 'Netflix')

print("--- Datenbank-Status ---")
if ergebnis:
    print(f"Gefundener Eintrag: {ergebnis[0]}")
else:
    print("Kein Eintrag gefunden.")

print(f"\nAlle Einträge im Speicher: {db.all()}")
```
