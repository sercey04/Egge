## 📘 Abschnitt A – Grundlagen: Was ist ein Ausdruck?

Ein **Ausdruck** ist eine Kombination aus:

- Werten (z. B. `5`, `3.5`)  
- Operatoren (z. B. `+`, `-`, `*`, `/`)  
- Variablen (z. B. `x`, `anzahl`)  

Python **berechnet** den Ausdruck und liefert ein Ergebnis.

Beispiele für Ausdrücke:

- `3 + 4`  
- `10 - 2 * 3`  
- `preis * anzahl`  


````python
# Probieren Sie einige Ausdrücke aus:
print(3 + 4)
print(10 - 2 * 3)
print((10 - 2) * 3)
````

## 📚 Glossar

- **Ausdruck** – Kombination aus Werten, Variablen und Operatoren, die zu einem Ergebnis führt.  
- **Operator** – Symbol für eine Operation (z. B. `+`, `-`, `*`, `/`, `//`, `%`, `**`).  
- **Rechenreihenfolge** – bestimmt, welcher Teil eines Ausdrucks zuerst berechnet wird.  
- **`input()`** – Funktion, mit der Benutzereingaben als Text eingelesen werden.  
- **`int()` / `float()`** – Funktionen zur Umwandlung von Text in Ganzzahlen bzw. Kommazahlen.  
- **`ZeroDivisionError`** – Fehler bei Division durch 0.  
- **`ValueError`** – Fehler, wenn eine Umwandlung (z. B. `int("abc")`) nicht möglich ist.
